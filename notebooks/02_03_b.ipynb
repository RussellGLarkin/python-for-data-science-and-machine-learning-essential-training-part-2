```python
# Visualize the distribution of the target variable, Survived, using a countplot.

sns.countplot(x='Survived', data=titanic_training, hue='Survived', palette='hls', legend=False)
plt.show()
```

```python
# Check for missing values.
#  - cabin has a significant number of missing values. 

titanic_training.isnull().sum()
```

```python
titanic_training.describe()
```

```python
# Clean data set to include features that are relevant to the prediction of survival. 

titanic_data = titanic_training.drop(['Name', 'Ticket', 'Cabin'], axis=1)
titanic_data.head()
```

```python
sns.boxplot(x='Parch', y='Age', data=titanic_data, palette='hls')
plt.show()
```

```python
# There were a lot of missing values in the Age column. Grouped the data by the number of parents/children aboard (Parch) and calculate the mean age for each group.

Parch_groups = titanic_data.groupby(titanic_data['Parch'])
Parch_groups.mean(numeric_only=True)
```

```python
# Fill missing values in the Age column based on the number of parents/children aboard (Parch).

def age_approx(cols):
  Age = cols.iloc[0]
  Parch = cols.iloc[1]
  if pd.isnull(Age):
    if Parch == 0:
      return 32
    elif Parch == 1:
      return 24
    elif Parch == 2:
      return 17
    elif Parch == 3:
      return 33
    elif Parch == 4:
      return 44
    else:
      return 30  # Close it out by returning the overall mean age for passengers (describe.mean(age))
  
  #: If the Age is not null, return the Age.
  else:
    return Age
```

```python
# Apply the age_approx function to fill in missing values in the Age column based on the number of parents/children aboard (Parch).

titanic_data['Age'] = titanic_data[['Age', 'Parch']].apply(age_approx, axis=1)

titanic_data.isnull().sum()
```

```python
titanic_data.describe()
```

```python
titanic_data.dropna(inplace=True)
titanic_data.reset_index(drop=True, inplace=True)

titanic_data.info()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

# Encode the gender column to convert the categorical values into numerical values.
gender_cat = titanic_data['Sex']
gender_encoded = label_encoder.fit_transform(gender_cat)

gender_encoded[:5]
```

```python
# Verify encoding by comparing first 5 rows of encoding to original data set. 
titanic_data.head(10)
```

```python
gender_df = pd.DataFrame(gender_encoded, columns=['Gender'])
gender_df.head()
```

```python
embarked_cat = titanic_data['Embarked']
embarked_encoded = label_encoder.fit_transform(embarked_cat)

embarked_encoded[:10]
```

```python
from sklearn.preprocessing import OneHotEncoder

binary_encoder = OneHotEncoder(categories='auto')
embarked_onehot = binary_encoder.fit_transform(embarked_encoded.reshape(-1, 1))
embarked_onehot_mat = embarked_onehot.toarray()
embarked_df = pd.DataFrame(embarked_onehot_mat, columns=['Embarked_C', 'Embarked_Q', 'Embarked_S'])
embarked_df.head()
```

```python
titanic_data.drop(['Sex', 'Embarked'], axis=1, inplace=True)
titanic_data.head()
```

```python
titanic_dummy = pd.concat([titanic_data, gender_df, embarked_df], axis=1).astype(float)
titanic_dummy.head()
```

```python
sns.heatmap(titanic_dummy.corr())
plt.show()
```

```python
titanic_dummy.drop(['Fare'], axis=1, inplace=True)
```

```python
titanic_dummy.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(titanic_dummy.drop('Survived', axis=1), 
                                                                       titanic_dummy['Survived'], test_size=0.2, 
                                                                       random_state=200)
```

```python
print(X_train.shape, y_train.shape)
```

```python
X_train[:5]
```

```python
LogReg = LogisticRegression(solver='liblinear')
LogReg.fit(X_train, y_train)
```

```python
ypred = LogReg.predict(X_test.values)
```

```python
print(classification_report(y_test, ypred))
```

```python
y_train_pred = cross_val_predict(LogReg, X_train, y_train, cv=5)
confusion_matrix(y_train, y_train_pred)
```

```python
precision_score(y_train, y_train_pred)
```
